```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
corpus="""
On Monday, November 23, 2026, Dr. Sarah Jenkins arrived in New York City at 8:30 a.m. to present her groundbreaking research on Natural Language Processing! She didn't expect over 500 researchers, data scientists, and software engineers to attend the keynote at Columbia University. 'It's truly remarkable how quickly modern algorithms are evolving,' she remarked, highlighting that running, runs, and ran all share a common linguistic root despite their differing tenses and contexts. 
"""
print(corpus)
```

##Tokenization
##Convert Peragraph to Sentenses
##Need to import below liberary

```python
import nltk
from nltk import sent_tokenize
nltk.download('punkt_tab')
sent_tokenize(corpus)
```

##Tokenization-2
##Peragraph/Sentense into words

```python
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

##WordPunct Tokenize

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordDetokenizer
Tokenizer=TreebankWordDetokenizer()
Tokenizer.tokenize(corpus)
```
