```python
words=["program", "programs", "programmer", "programming", "programmed"]
```

###PorterSteamer

```python
from nltk.stem import PorterStemmer
PS=PorterStemmer()
for word in words:
    print (word + " -------> " + PS.stem(word))
```

```python
PS.stem('congratulations')
```

###RegexStemer Class

```python
from nltk.stem import RegexpStemmer
RS=RegexpStemmer('ing$|s$', min=4)
RS.stem('eating')
```

##Morphological affixes---> Will match with the given word , if match remove that

```python
from nltk.stem import RegexpStemmer
RS=RegexpStemmer('$ing', min=4)
RS.stem('ingeating')
```

##Slowball Stemer

```python
from nltk.stem import SnowballStemmer
SS=SnowballStemmer('english')
for word in words:
    print(word+"----->"+SS.stem(word))
```
